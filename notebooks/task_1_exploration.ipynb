```python
new_observation = {
    "record_type": "observation",
    "pillar": "access",
    "indicator": "Account ownership (female)",
    "indicator_code": "ACC_OWN_FEMALE",
    "value_numeric": 29.0,
    "observation_date": "2021-01-01",
    "source_name": "World Bank Global Findex",
    "source_url": "https://www.worldbank.org/globalfindex",
    "confidence": "high",
    "collected_by": "Kifiya",
    "collection_date": "2026-01-28",
    "notes": "Used for gender gap analysis"
}

data = pd.concat([data, pd.DataFrame([new_observation])], ignore_index=True)
```

```python
new_event = {
    "record_type": "event",
    "category": "policy",
    "pillar": np.nan,
    "indicator": np.nan,
    "indicator_code": np.nan,
    "value_numeric": np.nan,
    "observation_date": "2020-12-01",
    "source_name": "National Bank of Ethiopia",
    "source_url": "https://nbe.gov.et",
    "confidence": "medium",
    "collected_by": "Kifiya",
    "collection_date": "2026-01-28",
    "notes": "Mobile money regulatory reform"
}

data = pd.concat([data, pd.DataFrame([new_event])], ignore_index=True)
```

```python
new_impact_link = {
    "parent_id": "EVENT_NBE_MM_POLICY_2020",
    "pillar": "access",
    "related_indicator": "ACC_MM_ACCOUNT",
    "impact_direction": "positive",
    "impact_magnitude": 0.15,
    "lag_months": 12,
    "evidence_basis": "Observed post-policy growth"
}

impact_links = pd.concat(
    [impact_links, pd.DataFrame([new_impact_link])],
    ignore_index=True
)
```

```python
# Save enriched datasets
data.to_csv(
    "../data/processed/ethiopia_fi_unified_data_enriched.csv",
    index=False
)

impact_links.to_csv(
    "../data/processed/impact_links_enriched.csv",
    index=False
)

print("Processed files saved successfully")
```
